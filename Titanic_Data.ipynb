```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\S.A Rathika Priyanka\Downloads\titanic_data.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df['Sex'].value_counts())
print(df['Survived'].value_counts())
print(df['Pclass'].value_counts())
```

```python
sns.histplot(df['Age'].dropna(), kde=True, bins=30)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age vs Survived')
plt.show()
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df)
plt.title('Age vs Fare (colored by Survival)')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['Survived','Pclass','Age','SibSp','Parch','Fare']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df[['Survived','Pclass','Age','Fare']], hue='Survived')
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival Count by Sex')
plt.show()
```

Summary of Findings:
1.About 38% of passengers survived overall.
2.Sex was the strongest survival factor — females had a much higher survival rate than males.
3.Passenger class (Pclass) mattered — 1st class passengers survived more than 3rd class.
4.Fare correlated with survival — higher fare passengers survived more, tying back to class.
5.Age alone didn't show a strong pattern, though very young children had slightly better survival odds.
6.Age has 177 missing values that would need handling in further analysis.
